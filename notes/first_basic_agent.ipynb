```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_groq import ChatGroq
from langchain.tools import tool
load_dotenv()
from langchain.agents import create_agent
llm = ChatGroq(
    model="openai/gpt-oss-20b",streaming=True
)



@tool
def addnums(a:int ,b:int ):
    """ it will return sum of two numbers
    args:
    a:number one
    b:number two """
    return a+b
#addnums.invoke({"a":12,"b":3})

@tool
def multiplynums(a:int ,b:int ):
    """ it will return product of two numbers
    args:
    a:number one
    b:number two """
    return a*b
#multiplynums.invoke({"a":12,"b":3})

```

Creating Agent

```python
agent = create_agent(
    model=llm,
    tools=[addnums,multiplynums],
    system_prompt="you are a match teacher and always use matchs for calculation"
)
```

Calling Agent

```python
res= agent.invoke({"messages":[{"role":"user","content":"what is 2+3 and product of 2 and 3" }]})
print(res["messages"][-1].content)
```
